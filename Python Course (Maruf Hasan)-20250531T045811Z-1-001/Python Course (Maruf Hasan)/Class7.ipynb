```python
NL = [23,45,67,66,92,34,69]
def isOdd(n):
  if n%2==1:
    return n

Odd_numbers =list( map(isOdd,NL))
print(Odd_numbers)
```

```python
NL = [23,45,67,66,92,34,69]
def isOdd(n):
  if n%2==1:
    return n

Odd_numbers =list(filter(lambda n: n%2==1,NL))
print(Odd_numbers)
```
