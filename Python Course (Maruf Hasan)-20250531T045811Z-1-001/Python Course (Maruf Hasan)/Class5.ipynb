```python
def set_username():
  username= (input("Enter The Username: "))
  for char in username:
    if char.isalpha() or char.isdigit():
      continue
    elif char==' ':
      print("Username is not ok")

    elif len(username)>=8 and len(username)<=20:
      print("User Is ok")

    else:
      print("Username is not Set")
set_username()
```
