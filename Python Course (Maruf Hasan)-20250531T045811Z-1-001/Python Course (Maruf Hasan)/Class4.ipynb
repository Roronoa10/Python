```python
plaintext = input("Please enter: ")
key = int(input("Enter The Key: "))
cipher=""
for char in plaintext:
    cipher += chr(ord(char)+key)
print("Cipher: ", cipher)

messege=""
for char in cipher:
    messege += chr(ord(char)-key)
print("messege: ", messege)
```

```python
import math
n= int (input("put a number: "))
prime = True
for i in range(2,int(math.sqrt(n)),1):
  if n % i ==0:
    prime= False
    break
  if prime is True:
   print(n,'is a prime number')
  else:
     print(n, 'is not a prime number')
```
