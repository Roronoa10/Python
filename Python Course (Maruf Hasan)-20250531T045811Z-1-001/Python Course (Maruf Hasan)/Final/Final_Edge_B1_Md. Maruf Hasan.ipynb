```python
import pandas as pd

df = pd.read_csv('/content/sample_data/employees_large.csv')

avg_salary_by_dept = df.groupby('Department')['Salary'].mean()

print("Average Salary by Department:")
print(avg_salary_by_dept)

department_with_highest_salary = avg_salary_by_dept.idxmax()
highest_salary = avg_salary_by_dept.max()

print(f"\nDepartment with the highest average salary: {department_with_highest_salary}")
print(f"Highest average salary: ${highest_salary}")
```

```python
import pandas as pd

sales_df = pd.read_csv('/content/sample_data/sales_large.csv')

total_sales_by_product_region = sales_df.groupby(['Region', 'Product'])['Sales'].sum().reset_index()

highest_sales_product = total_sales_by_product_region.groupby('Product')['Sales'].sum().idxmax()
print(f"Product with the highest total sales across all regions: {highest_sales_product}")
```

```python
import numpy as np

A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
C_add = np.add(A, B)
C_sub = np.subtract(A, B)
C_mul = np.dot(A, B)
C_elem_mul = np.multiply(A, B)
print("Matrix A:\n", A)
print("Matrix B:\n", B)
print("Matrix A + B:\n", C_add)
print("Matrix A - B:\n", C_sub)
print("Matrix A * B (Matrix multiplication):\n", C_mul)
print("Element-wise multiplication A * B:\n", C_elem_mul)
```

```python
import numpy as np

data = np.random.randn(10000)
mean_data = np.mean(data)
median_data = np.median(data)
std_dev_data = np.std(data)
min_data = np.min(data)
max_data = np.max(data)
greater_than_mean = np.sum(data > mean_data)
normalized_data = (data - min_data) / (max_data - min_data)

print(f"Mean: {mean_data}, Median: {median_data}, Standard Deviation: {std_dev_data}")
print(f"Min value: {min_data}, Max value: {max_data}")
print(f"Number of values greater than the mean: {greater_than_mean}")
print(f"Normalized dataset (first 10 values): {normalized_data[:10]}")
```

```python
temperatures = [28, 32, 35, 31, 29, 30, 33, 36, 27, 25, 34, 30, 29, 37, 38, 26, 31, 35, 33, 32, 36, 34, 29, 28, 27, 35, 32, 30, 31, 33]

above_32 = [temp for temp in temperatures if temp > 32]

classification = [(temp, 'Hot' if temp > 35 else 'Warm' if temp >= 30 else 'Cool') for temp in temperatures]
avg_temp = np.mean(temperatures)
temp_diffs = [temp - avg_temp for temp in temperatures]

print(f"Temperatures above 32°C: {above_32}")
print(f"Temperature classifications: {classification}")
print(f"Temperature differences from the average: {temp_diffs}")
```
